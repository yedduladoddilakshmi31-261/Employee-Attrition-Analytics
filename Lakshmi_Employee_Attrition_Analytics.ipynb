```python
department_attrition = pd.crosstab(
    df['Department'],
    df['Attrition'],
    normalize='index'
) * 100

print(department_attrition.round(2))
```

```python
overtime_attrition = pd.crosstab(
    df['OverTime'],
    df['Attrition'],
    normalize='index'
) * 100

print(overtime_attrition.round(2))
```

```python
jobrole_attrition = pd.crosstab(
    df['JobRole'],
    df['Attrition'],
    normalize='index'
) * 100

print(jobrole_attrition.round(2))
```

```python
print("Total Employees:", len(df))
print("Employees Who Left:", (df['Attrition'] == 'Yes').sum())
print("Employees Who Stayed:", (df['Attrition'] == 'No').sum())

attrition_rate = (df['Attrition'] == 'Yes').mean() * 100
print(f"Overall Attrition Rate: {attrition_rate:.2f}%")
```

```python
df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=[18, 25, 35, 45, 60],
    labels=['18-25', '26-35', '36-45', '46-60']
)

print(df['AgeGroup'].value_counts().sort_index())
```

```python
age_attrition = pd.crosstab(
    df['AgeGroup'],
    df['Attrition'],
    normalize='index'
) * 100

print(age_attrition.round(2))
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='AgeGroup', hue='Attrition')

plt.title('Employee Attrition by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')

plt.show()
```
